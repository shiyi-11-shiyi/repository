```python
class TreeNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None


def array_to_tree(array):
    if not array or array[0] is None:
        return None

    # 创建根节点
    root = TreeNode(array[0])
    queue = [root]  # 用队列进行层级构造
    i = 1           # 数组下标指针

    while queue and i < len(array):
        current = queue.pop(0)  # 当前节点出队

        # 构造左子节点
        if array[i] is not None:
            current.left = TreeNode(array[i])
            queue.append(current.left)
        i += 1

        # 构造右子节点
        if i < len(array) and array[i] is not None:
            current.right = TreeNode(array[i])
            queue.append(current.right)
        i += 1

    return root


def print_tree(node, level=0, prefix="Root: "):
    if node:
        print(" " * (level * 4) + prefix + str(node.key))
        print_tree(node.left, level + 1, "L--- ")
        print_tree(node.right, level + 1, "R--- ")
```

```python
# 测试代码
if __name__ == "__main__":
    array = [10, 5, 15, 3, 7, None, 20]
    tree = array_to_tree(array)

    print("构造的层级结构二叉树：")
    print_tree(tree)
```
