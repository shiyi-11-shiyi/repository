```python
class Node:
    """链表节点类"""
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    """单链表类"""
    def __init__(self):
        self.head = None

    # --- 辅助方法：在末尾添加节点 (用于构建测试数据) ---
    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        last_node = self.head
        while last_node.next:
            last_node = last_node.next
        last_node.next = new_node

    # --- 辅助方法：打印链表 ---
    def display(self):
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        print(" -> ".join(elements))

    def delete_at_index(self, index):
        """
        根据索引删除节点
        例如: [a, b, c, d] 删除索引 2 ("c")
        """
        # 1. 如果链表为空
        if not self.head:
            print("链表为空，无法删除")
            return

        # 2. 如果要删除的是头节点 (index 0)
        if index == 0:
            self.head = self.head.next
            print(f"已删除头节点")
            return

        # 3. 寻找待删除节点的前一个节点 (prev_node)
        current = self.head
        count = 0
        
        # 我们需要走到 index - 1 的位置
        while current.next and count < index - 1:
            current = current.next
            count += 1
        
        # 检查索引是否越界 (current.next 为 None 说明 index 太大了)
        if not current.next:
            print(f"索引 {index} 越界")
            return

        # 4. 执行删除操作 (对应PPT上的步骤)
        # PPT步骤: 让 ["b"] 的 next 指向 ["d"]
        node_to_delete = current.next
        current.next = node_to_delete.next
        
        # PPT步骤: ["c"] 不再被引用 (Python会自动回收内存)
        print(f"已删除索引 {index} 的节点: {node_to_delete.data}")

    def remove(self, data):
        """
        根据值删除节点 (补充功能)
        例如: 删除 "c"
        """
        current = self.head

        # 1. 如果头节点就是要删的值
        if current and current.data == data:
            self.head = current.next
            print(f"已删除节点: {data}")
            return

        # 2. 搜索要删除的节点
        while current.next:
            if current.next.data == data:
                # 找到目标，执行删除
                current.next = current.next.next
                print(f"已删除节点: {data}")
                return
            current = current.next
        
        print(f"未找到节点: {data}")

if __name__ == "__main__":
    print(" 删除索引 2 ('c')")
    
    # 1. 构建链表: a -> b -> c -> d
    ll = LinkedList()
    ll.append('a')
    ll.append('b')
    ll.append('c')
    ll.append('d')
    
    print("删除前:")
    ll.display()  # 输出: a -> b -> c -> d
    
    # 2. 执行删除: 索引 2 是 'c'
    # 逻辑: 找到索引1 ('b') -> 让 'b'.next 指向 'd' -> 'c' 被丢弃
    ll.delete_at_index(2)
    
    print("删除后:")
    ll.display()  # 输出: a -> b -> d

    print("\n按值删除 'a' (头节点)")
    ll.remove('a')
    ll.display()  # 输出: b -> d
```
