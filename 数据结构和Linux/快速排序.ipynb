```python
import random
from typing import List

def randomized_quick_sort(arr: List[int], low: int, high: int) -> None:
    """随机化快速排序"""
    if low < high:
        # 随机选择轴元素并交换到末尾
        random_index = random.randint(low, high)
        arr[random_index], arr[high] = arr[high], arr[random_index]
        pivot = arr[high]
        
        # 标准 Lomuto 划分
        i = low - 1
        for j in range(low, high):
            if arr[j] <= pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]
        
        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        pivot_index = i + 1
        
        # 递归排序左右两部分
        randomized_quick_sort(arr, low, pivot_index - 1)
        randomized_quick_sort(arr, pivot_index + 1, high)

def three_way_quick_sort(arr: List[int], low: int, high: int) -> None:
    """三路快速排序"""
    if low >= high:
        return
    
    # 随机选择轴并交换到开头
    random_index = random.randint(low, high)
    arr[low], arr[random_index] = arr[random_index], arr[low]
    pivot = arr[low]
    
    lt = low     # 小于轴的边界
    gt = high    # 大于轴的边界
    i = low + 1  # 当前遍历指针
    
    while i <= gt:
        if arr[i] < pivot:
            arr[i], arr[lt] = arr[lt], arr[i]
            lt += 1
            i += 1
        elif arr[i] > pivot:
            arr[i], arr[gt] = arr[gt], arr[i]
            gt -= 1
        else:  # arr[i] == pivot
            i += 1
    
    # 递归排序小于和大于部分
    three_way_quick_sort(arr, low, lt - 1)
    three_way_quick_sort(arr, gt + 1, high)

def test_sorts():
    """测试函数"""
    print("快速排序优化测试 ")
    
    # 测试随机化快速排序
    print("\n1. 随机化快速排序测试:")
    arr1 = [3, 7, 8, 5, 2, 1, 9, 5, 4]
    print(f"原始数组: {arr1}")
    randomized_quick_sort(arr1, 0, len(arr1) - 1)
    print(f"排序结果: {arr1}")
    
    # 测试三路快速排序
    print("\n2. 三路快速排序测试:")
    arr2 = [3, 7, 5, 5, 2, 5, 9, 5, 1, 5, 4, 5]
    print(f"原始数组: {arr2}")
    three_way_quick_sort(arr2, 0, len(arr2) - 1)
    print(f"排序结果: {arr2}")
    
    # 性能对比测试
    print("\n3. 性能对比测试:")
    
    # 生成包含重复元素的测试数组
    test_arr = [random.randint(0, 10) for _ in range(20)]
    print(f"测试数组（20个元素，值范围0-10）: {test_arr}")
    
    # 使用随机化快速排序
    arr_rqs = test_arr.copy()
    randomized_quick_sort(arr_rqs, 0, len(arr_rqs) - 1)
    print(f"随机化快速排序结果: {arr_rqs}")
    
    # 使用三路快速排序
    arr_threeway = test_arr.copy()
    three_way_quick_sort(arr_threeway, 0, len(arr_threeway) - 1)
    print(f"三路快速排序结果: {arr_threeway}")
    
    # 验证排序正确性
    assert sorted(test_arr) == arr_rqs == arr_threeway
    print("\n✓ 两种排序算法结果都正确！")

if __name__ == "__main__":
    test_sorts()
```
