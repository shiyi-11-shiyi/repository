```python
class TreeNode:
    def __init__(self, key):
        self.key = key
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None

    # 插入节点
    def insert(self, node, key):
        if not node:
            return TreeNode(key)
        if key < node.key:
            node.left = self.insert(node.left, key)
        else:
            node.right = self.insert(node.right, key)
        return node

    # 构造 BST
    def build_bst(self, values):
        for value in values:
            self.root = self.insert(self.root, value)

    # 查找中序前驱
    def find_inorder_predecessor(self, node):
        current = node.left
        while current and current.right:
            current = current.right
        return current

    # 查找中序后继
    def find_inorder_successor(self, node):
        current = node.right
        while current and current.left:
            current = current.left
        return current

    # 删除节点
    def delete(self, node, key, use_predecessor=True):
        if not node:
            return node

        if key < node.key:
            node.left = self.delete(node.left, key, use_predecessor)
        elif key > node.key:
            node.right = self.delete(node.right, key, use_predecessor)
        else:  # 找到要删除的节点
            # 无子节点
            if not node.left and not node.right:
                return None

            # 只有右子树
            if not node.left:
                return node.right

            # 只有左子树
            if not node.right:
                return node.left

            # 两个子树均存在
            if use_predecessor:  # 使用中序前驱
                predecessor = self.find_inorder_predecessor(node)
                node.key = predecessor.key
                node.left = self.delete(node.left, predecessor.key, use_predecessor)
            else:  # 使用中序后继
                successor = self.find_inorder_successor(node)
                node.key = successor.key
                node.right = self.delete(node.right, successor.key, use_predecessor)

        return node

    # 打印树结构
    def print_tree(self, node, level=0, prefix="Root: "):
        if node:
            print(" " * (level * 4) + prefix + str(node.key))
            if node.left or node.right:
                self.print_tree(node.left, level + 1, "L--- ")
                self.print_tree(node.right, level + 1, "R--- ")
```

```python
# 测试代码
if __name__ == "__main__":
    # 构建 BST
    bst = BST()
    values = [50, 30, 70, 20, 40, 60, 80]
    bst.build_bst(values)

    print("构建 BST：")
    bst.print_tree(bst.root)

    # 删除节点使用中序前驱法
    print("\n删除根节点 50（中序前驱法）：")
    bst.root = bst.delete(bst.root, 50, use_predecessor=True)
    bst.print_tree(bst.root)

    # 构建原始 BST 再删除节点使用中序后继法
    bst = BST()
    bst.build_bst(values)

    print("\n删除根节点 50（中序后继法）：")
    bst.root = bst.delete(bst.root, 50, use_predecessor=False)
    bst.print_tree(bst.root)
```
